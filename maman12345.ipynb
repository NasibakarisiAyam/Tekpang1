```python
#hubungan antara karbohidrat dan kalori: karbohidrat sumbu x, kalori sumbu y
x = df['Carbohydrates']
y = df['Calories']

#melihat hubungan x dan y di grafik
#menggunakan library untuk grafik matplotlib
import matplotlib.pyplot as plt
plt.scatter(x,y)
plt.title("hubungan karbohidrat dan kalori")
plt.xlabel("Karbohidrat")
plt.ylabel("Kalori")
plt.show()
```

```python
from scipy import stats #library machine learning
```

```python
#y = mx + c    m : gradien / kemiringan
#c : konstanta
m, c, r, p, stderr = stats.linregress(x,y)
```

```python
#untuk mengetahui gradien
m
```

```python
c
```

```python
# y = m*x + c
#misal karbohidrat x=100 gram
ypred = m*100 + c
ypred
```

```python

def myfunc(x):
  return m * x + c

mymodel = list(map(myfunc, x))
plt.scatter(x,y)
plt.plot(x, mymodel, color='red')
plt.title("hubungan karbohidrat dan kalori")
plt.xlabel("Karbohidrat")
plt.ylabel("Kalori")
plt.show()
```
